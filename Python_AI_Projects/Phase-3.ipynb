# Vector operations

```python
import numpy as np
import matplotlib.pyplot as plt


Vector1 = np.array([1, 2, 3])
Vector2 = np.array([[4], [5], [6]])
print("Vector1:")
print(Vector1)  
print("Vector2:")
print(Vector2)
print("Shape of Vector1:", Vector1.shape)
print("Shape of Vector2:", Vector2.shape)

#Vector addition
Vector3 = Vector1 + Vector2.flatten()  # Flatten Vector2 to make it compatible  

print("Vector1 + Vector2:")
print(Vector3)

#Dot product
Dot_Product = np.dot(Vector1, Vector2.flatten())  # Flatten Vector2 for dot product
print("Dot Product of Vector1 and Vector2:")
print(Dot_Product)

#Cross product
Cross_Product = np.cross(Vector1, Vector2.flatten())  # Flatten Vector2 for cross   product
print("Cross Product of Vector1 and Vector2:")
print(Cross_Product)        

#multiplication of vectors
Elementwise_Product = Vector1 * Vector2.flatten()  # Flatten Vector2 for element-wise multiplication
print("Element-wise Product of Vector1 and Vector2:")       
print(Elementwise_Product)              

multiplication = np.multiply(Vector1, Vector2.flatten())  # Flatten Vector2 for multiplication
print("Multiplication of Vector1 and Vector2:")
print(multiplication)  

angle = np.arccos(np.dot(Vector1, Vector2.flatten()) / (np.linalg.norm(Vector1) * np.linalg.norm(Vector2.flatten())))  # Flatten Vector2 for angle calculation
print("Angle between Vector1 and Vector2 (in radians):")        
print(angle)    


foodTypes = np.array(["Pizza", "Burger", "Sushi", "Pasta"])
print("Food Types:", foodTypes)

vectorized_upper = np.vectorize(str.upper)
upper_foodTypes = vectorized_upper(foodTypes)
print("Uppercase Food Types:", upper_foodTypes)

# 5*5 matrix with values from 1 to 25
matrix = np.arange(1, 26).reshape(5, 5)
print("5x5 Matrix:")
print(matrix)

# divide each element by 5
divided_matrix = matrix / 5         
print("Matrix divided by 5:")
print(divided_matrix)   
```
