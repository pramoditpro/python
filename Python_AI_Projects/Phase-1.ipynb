### Vector Metrices Tensors

```python
vector_a = np.array([1, 2, 3.3])

matrix_b = np.array([[4, 5, 6], [7, 8, 9]])

tensor_c = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])

print("Vector (1D Array):", vector_a)
print("Matrix (2D Array):\n", matrix_b)     
print("Tensor (3D Array):\n", tensor_c)

### Array properties
print("Shape of vector_a:", vector_a.shape) 
print("Shape of matrix_b:", matrix_b.shape)
print("Shape of tensor_c:", tensor_c.shape)
print("Dimensions of vector_a:", vector_a.ndim) 
print("Dimensions of tensor_c:", tensor_c.ndim)
print("Size of vector_a:", vector_a.size) 
print("Size of tensor_c:", tensor_c.size)
print("Data type of vector_a:", vector_a.dtype) 
print("Data type of tensor_c:", tensor_c.dtype)


reshape_matrix = matrix_b.reshape(3, 2)
print("Reshaped Matrix (3x2):\n", reshape_matrix)  

flattened_tensor = tensor_c.flatten()
print("Flattened Tensor:\n", flattened_tensor)

transposed_matrix = matrix_b.T
print("Transposed Matrix:\n", transposed_matrix)    

ravelled_matrix = reshape_matrix.ravel()
print("Ravelled Matrix:\n", ravelled_matrix)
    
```
